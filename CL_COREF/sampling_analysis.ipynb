```python
def process_samples(samples, mentions):
    positive_entities_same_subtopic = []
    positive_entities_different_subtopic = []
    positive_similarity_scores = []

    negative_entities_same_subtopic = []
    negative_entities_different_subtopic = []
    negative_similarity_scores = []

    for pair_key, (label, score) in samples.items():
        key1, key2 = pair_key.split("###")
        m1, m2 = mentions[key1], mentions[key2]

        if label == 1:
            positive_similarity_scores.append(score)
            if m1.subtopic == m2.subtopic:
                positive_entities_same_subtopic.append(pair_key)
            else:
                positive_entities_different_subtopic.append(pair_key)
        else:
            negative_similarity_scores.append(score)
            if m1.subtopic == m2.subtopic:
                negative_entities_same_subtopic.append(pair_key)
            else:
                negative_entities_different_subtopic.append(pair_key)

    total_samples = len(samples)
    total_positives = len(positive_entities_same_subtopic) + len(positive_entities_different_subtopic)
    total_negatives = len(negative_entities_same_subtopic) + len(negative_entities_different_subtopic)

    print("Number of pairs:", total_samples)
    print("Number of positive:", total_positives)
    print("Number of negatives:", total_negatives)
    print("Number of same Subtopic:",  len(positive_entities_same_subtopic) + len(negative_entities_same_subtopic))
    print("Number of different Subtopic:",  len(positive_entities_different_subtopic) + len(negative_entities_different_subtopic))
    print("Fraction positive:", total_positives / total_samples)
    print("Fraction same subtopic:", (len(positive_entities_same_subtopic) + len(negative_entities_same_subtopic)) / total_samples)
    print("Median pos. similarity score:", statistics.median(positive_similarity_scores))
    print("Median neg. similarity score:", statistics.median(negative_similarity_scores))



```

```python
def samples_by_topic(samples, mentions):
    positive_samples_by_topic = {}
    negative_samples_by_topic = {}
    for pair_key, (label, score) in samples.items():
        key1, key2 = pair_key.split("###")
        m1, m2 = mentions[key1], mentions[key2]
        #mentions belong to the same topic, throw exception otherwise
        if m1.topic!=m2.topic:
            raise Exception("mentions from different topic")
        
        if label==1:
            if m1.topic in positive_samples_by_topic:
                positive_samples_by_topic[m1.topic].append(pair_key)
            else:
                positive_samples_by_topic[m1.topic] = [key1]
        else:
            if m1.topic in negative_samples_by_topic:
                negative_samples_by_topic[m1.topic].append(pair_key)
            else:
                negative_samples_by_topic[m1.topic] = [key1]

    return positive_samples_by_topic, negative_samples_by_topic 
```

```python
# Mentions by topic + samples per topic
total = 0 
for k,v in cdcr_train.mentions_by_topic.items():


    topic_id = k.split("-")[1]
    if topic_id not in positive_by_topic:
        continue

    mentions_in_this_topic = [mention for mention in v if mention in cdcr_train.mentions_by_type["event"]]
    pos_limit =  6*len(mentions_in_this_topic)**(1/2)
    neg_limit = pos_limit*20
    pos_samples = len(positive_by_topic[topic_id])
    neg_samples = len(negative_by_topic[topic_id])
    
    total+= pos_limit

    if pos_samples>pos_limit:
        print(f"topic {topic_id} num mentions {len(mentions_in_this_topic)} pos limit = {round(pos_limit)} neg limit = {round(neg_limit)} positive samples {pos_samples} negative samples {neg_samples} ")
print(total)
```

```python
import random

def sample_key_value_pairs(dictionary, sample_size):
    if len(dictionary) <= sample_size:
        return dict(dictionary)

    sampled_keys = random.sample(dictionary.keys(), sample_size)
    sampled_pairs = {key: dictionary[key] for key in sampled_keys}
    return sampled_pairs

# Example usage
my_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

max_sample_size = 2
sampled_dict = sample_key_value_pairs(my_dict, max_sample_size)

print("Original Dictionary:", my_dict)
print("Sampled Dictionary:", sampled_dict)
```
