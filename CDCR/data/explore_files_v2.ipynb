```python
config_file = ecbplus_config
train = dataset.CrossEncoderDatasetFull(config=config_file,split_name='train')
test = dataset.CrossEncoderDatasetFull(config=config_file,split_name='test')
val = dataset.CrossEncoderDatasetFull(config=config_file,split_name='val')
```

```python
def check_coreference_clusters(mentions_list):
    corefs_temp = [mention["coref_chain"] for mention in mentions_list]
    counter_temp = collections.Counter(corefs_temp)
    counter_mean = sum(counter_temp.values())/len(counter_temp)
    singletons_by_count = sum(1 for count in counter_temp.values() if count == 1)
    singletons_by_name = sum(1 for key in counter_temp.keys() if "Singleton" in key)
    singletons_string = f"{singletons_by_count} singletons by count | {singletons_by_name} by name"
    coreference_string = f"| {len(counter_temp)-singletons_by_count} Non-singletons | total {len(counter_temp)} clusters"
    chain_averages = f"| {counter_mean} mentions per cluster"
    return singletons_string + coreference_string + chain_averages
```

```python
def check_topics(dataset,split):
    len_subtopics = len(set([mention["subtopic_id"] for mention in dataset.mentions]))
    return f"the {split} dataset has {len(dataset.mentions_by_topics.keys())} topics and {len_subtopics} subtopics and {len(dataset.first)} combinations after undersampling"

def check_documents(dataset,split):
    documents_list = set([mention["conll_doc_key"] for mention in dataset.mentions])
    list_tokens = [len(dataset.documents[document_key]) for document_key in documents_list]
    return f"the {split} dataset has {len(documents_list)} documents and {sum(list_tokens)} tokens "

def coreference_stats(dataset,split):
    documents_list = set([mention["conll_doc_key"] for mention in dataset.mentions])
    list_tokens = [len(dataset.documents[document_key]) for document_key in documents_list]
    return f"the {split} dataset has {len(documents_list)} documents and {sum(list_tokens)} tokens "
```
