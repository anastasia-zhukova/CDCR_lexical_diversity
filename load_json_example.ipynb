# Load and Explore WECEng Event Mentions

This notebook demonstrates how to load and analyze the WECEng event mentions JSON file.

## Setup

```python
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter, defaultdict

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)

print("Libraries loaded successfully!")
```


## Load JSON Data

```python
# Load the JSON file
json_path = 'CL_COREF/data/CD2CR/train/entity_mentions.json'

print(f"Loading: {json_path}")
with open(json_path, 'r') as f:
    mentions = json.load(f)

print(f"✓ Loaded {len(mentions):,} event mentions")
print(f"\nData type: {type(mentions)}")
print(f"First mention type: {type(mentions[0])}")
```

## Explore First Few Mentions

```python
# Display first mention in detail
print("First mention (formatted):\n")
print(json.dumps(mentions[0], indent=2))
```

```python
# Check what keys are available in mentions
print("Available keys in a mention:")
print(list(mentions[0].keys()))
```

## Convert to DataFrame for Easy Analysis

```python
# Convert to DataFrame
df = pd.DataFrame(mentions)

print(f"DataFrame shape: {df.shape}")
print(f"Columns: {len(df.columns)}")
print(f"\nColumn names:")
print(list(df.columns))

# Display first few rows
df.topic.unique()
```

## Basic Statistics

```python
print("="*70)
print("BASIC STATISTICS")
print("="*70)

print(f"\nTotal Mentions: {len(df):,}")
print(f"\nUnique Topics: {df['topic'].nunique()}")
print(f"Topic IDs: {sorted(df['topic'].unique())}")

print(f"\nUnique Subtopics: {df['subtopic_id'].nunique() if 'subtopic_id' in df.columns else 'N/A'}")
print(f"Unique Documents: {df['doc_id'].nunique() if 'doc_id' in df.columns else 'N/A'}")
print(f"Unique Coreference Chains: {df['coref_chain'].nunique()}")

if 'mention_type' in df.columns:
    print(f"\nMention Types:")
    print(df['mention_type'].value_counts())

if 'is_singleton' in df.columns:
    print(f"\nSingletons: {df['is_singleton'].sum():,} ({df['is_singleton'].sum()/len(df)*100:.1f}%)")
    print(f"Non-singletons: {(~df['is_singleton']).sum():,} ({(~df['is_singleton']).sum()/len(df)*100:.1f}%)")
```

## Topic Distribution

```python
# Mentions per topic
topic_counts = df['topic_id'].value_counts().sort_index()

print("Mentions per Topic:")
print(topic_counts)

# Visualize
fig, ax = plt.subplots(figsize=(14, 6))
topic_counts.plot(kind='bar', ax=ax, color='steelblue', alpha=0.7)
ax.set_xlabel('Topic ID', fontsize=12, fontweight='bold')
ax.set_ylabel('Number of Mentions', fontsize=12, fontweight='bold')
ax.set_title('Mentions Distribution per Topic', fontsize=14, fontweight='bold')
ax.grid(axis='y', alpha=0.3)

# Add value labels
for i, v in enumerate(topic_counts):
    ax.text(i, v, str(v), ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

## Coreference Chain Analysis

```python
# Cluster sizes (mentions per coreference chain)
cluster_sizes = df['coref_chain'].value_counts()

print("="*70)
print("COREFERENCE CHAIN STATISTICS")
print("="*70)

print(f"\nTotal Clusters: {len(cluster_sizes):,}")
print(f"Singletons: {(cluster_sizes == 1).sum():,}")
print(f"Non-singleton clusters: {(cluster_sizes > 1).sum():,}")
print(f"\nCluster Size Statistics:")
print(f"  Mean: {cluster_sizes.mean():.2f}")
print(f"  Median: {cluster_sizes.median():.0f}")
print(f"  Min: {cluster_sizes.min()}")
print(f"  Max: {cluster_sizes.max()}")
print(f"  Std: {cluster_sizes.std():.2f}")

print(f"\nTop 10 Largest Clusters:")
print(cluster_sizes.head(10))
```

```python
# Visualize cluster size distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Histogram of cluster sizes
cluster_sizes.hist(bins=50, ax=ax1, color='coral', alpha=0.7, edgecolor='black')
ax1.set_xlabel('Cluster Size (mentions per chain)', fontsize=11, fontweight='bold')
ax1.set_ylabel('Frequency', fontsize=11, fontweight='bold')
ax1.set_title('Distribution of Cluster Sizes', fontsize=12, fontweight='bold')
ax1.grid(axis='y', alpha=0.3)
ax1.axvline(cluster_sizes.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {cluster_sizes.mean():.1f}')
ax1.axvline(cluster_sizes.median(), color='green', linestyle='--', linewidth=2, label=f'Median: {cluster_sizes.median():.0f}')
ax1.legend()

# Bar chart of cluster size categories
size_categories = pd.cut(cluster_sizes, bins=[0, 1, 2, 5, 10, 20, 50, float('inf')],
                         labels=['1 (singleton)', '2', '3-5', '6-10', '11-20', '21-50', '50+'])
size_counts = size_categories.value_counts().sort_index()

size_counts.plot(kind='bar', ax=ax2, color='purple', alpha=0.7)
ax2.set_xlabel('Cluster Size Range', fontsize=11, fontweight='bold')
ax2.set_ylabel('Number of Clusters', fontsize=11, fontweight='bold')
ax2.set_title('Clusters by Size Category', fontsize=12, fontweight='bold')
ax2.tick_params(axis='x', rotation=45)
ax2.grid(axis='y', alpha=0.3)

# Add value labels
for i, v in enumerate(size_counts):
    ax2.text(i, v, str(v), ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

## Mention Text Analysis

```python
# Analyze mention text if available
if 'tokens_str' in df.columns:
    print("Sample mention texts:")
    print(df['tokens_str'].head(10).to_list())
    
    # Mention length analysis
    if 'tokens_text' in df.columns:
        df['mention_length'] = df['tokens_text'].apply(len)
        
        print(f"\nMention Length Statistics:")
        print(df['mention_length'].describe())
        
        # Plot distribution
        fig, ax = plt.subplots(figsize=(12, 6))
        df['mention_length'].hist(bins=30, ax=ax, color='teal', alpha=0.7, edgecolor='black')
        ax.set_xlabel('Mention Length (tokens)', fontsize=11, fontweight='bold')
        ax.set_ylabel('Frequency', fontsize=11, fontweight='bold')
        ax.set_title('Distribution of Mention Lengths', fontsize=12, fontweight='bold')
        ax.grid(axis='y', alpha=0.3)
        plt.tight_layout()
        plt.show()
elif 'tokens_text' in df.columns:
    print("Sample tokens:")
    print(df['tokens_text'].head(10).to_list())
```

## Topic-wise Cluster Analysis

```python
# Analyze clusters per topic
topic_cluster_stats = []

for topic_id in sorted(df['topic_id'].unique()):
    topic_df = df[df['topic_id'] == topic_id]
    topic_clusters = topic_df['coref_chain'].value_counts()
    
    topic_cluster_stats.append({
        'topic_id': topic_id,
        'mentions': len(topic_df),
        'clusters': len(topic_clusters),
        'singletons': (topic_clusters == 1).sum(),
        'non_singletons': (topic_clusters > 1).sum(),
        'avg_cluster_size': topic_clusters.mean(),
        'max_cluster_size': topic_clusters.max()
    })

topic_stats_df = pd.DataFrame(topic_cluster_stats)

print("\nPer-Topic Statistics:")
print(topic_stats_df.to_string(index=False))
```

```python
# Visualize topic statistics
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Mentions per topic
axes[0, 0].bar(topic_stats_df['topic_id'].astype(str), topic_stats_df['mentions'], color='steelblue', alpha=0.7)
axes[0, 0].set_xlabel('Topic ID', fontweight='bold')
axes[0, 0].set_ylabel('Mentions', fontweight='bold')
axes[0, 0].set_title('Mentions per Topic', fontweight='bold')
axes[0, 0].grid(axis='y', alpha=0.3)

# Clusters per topic
axes[0, 1].bar(topic_stats_df['topic_id'].astype(str), topic_stats_df['clusters'], color='coral', alpha=0.7)
axes[0, 1].set_xlabel('Topic ID', fontweight='bold')
axes[0, 1].set_ylabel('Clusters', fontweight='bold')
axes[0, 1].set_title('Clusters per Topic', fontweight='bold')
axes[0, 1].grid(axis='y', alpha=0.3)

# Average cluster size per topic
axes[1, 0].bar(topic_stats_df['topic_id'].astype(str), topic_stats_df['avg_cluster_size'], color='purple', alpha=0.7)
axes[1, 0].set_xlabel('Topic ID', fontweight='bold')
axes[1, 0].set_ylabel('Avg Cluster Size', fontweight='bold')
axes[1, 0].set_title('Average Cluster Size per Topic', fontweight='bold')
axes[1, 0].grid(axis='y', alpha=0.3)

# Max cluster size per topic
axes[1, 1].bar(topic_stats_df['topic_id'].astype(str), topic_stats_df['max_cluster_size'], color='green', alpha=0.7)
axes[1, 1].set_xlabel('Topic ID', fontweight='bold')
axes[1, 1].set_ylabel('Max Cluster Size', fontweight='bold')
axes[1, 1].set_title('Maximum Cluster Size per Topic', fontweight='bold')
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Examine Specific Examples

```python
# Get the largest cluster
largest_cluster = cluster_sizes.idxmax()
largest_cluster_mentions = df[df['coref_chain'] == largest_cluster]

print(f"Largest Cluster: {largest_cluster}")
print(f"Size: {len(largest_cluster_mentions)} mentions")
print(f"\nMentions in this cluster:")

for idx, row in largest_cluster_mentions.head(10).iterrows():
    mention_text = row.get('tokens_str', row.get('mention_id', 'N/A'))
    topic = row.get('topic_id', 'N/A')
    doc = row.get('doc_id', 'N/A')
    print(f"  [{topic}/{doc}] {mention_text}")

if len(largest_cluster_mentions) > 10:
    print(f"  ... and {len(largest_cluster_mentions) - 10} more")
```

```python
# Sample a few mentions from different topics
print("\nSample mentions from each topic:\n")

for topic_id in sorted(df['topic_id'].unique())[:5]:  # First 5 topics
    topic_mentions = df[df['topic_id'] == topic_id]
    print(f"Topic {topic_id} ({len(topic_mentions)} mentions):")
    
    for idx, row in topic_mentions.head(3).iterrows():
        mention_text = row.get('tokens_str', row.get('mention_id', 'N/A'))
        chain = row.get('coref_chain', 'N/A')
        print(f"  - {mention_text} [chain: {chain}]")
    print()
```
